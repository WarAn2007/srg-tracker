# 02 — Baseline and candidate models

Training models with the `src.train`. Saving the best model by testing them with `validation`. `Test` is not used here.

```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT))

import pandas as pd
from src.config import METRICS_DIR

comparison_path = METRICS_DIR / 'validation_model_comparison.csv'
if not comparison_path.exists():
    raise FileNotFoundError('First complete: python -m src.train')
comparison = pd.read_csv(comparison_path)
comparison
```

Metrics to evaluate: min `RMSE` for score, max `F1` of class `enroll` for outcome, max `Macro-F1` for pace.

```python
winners = {
    'score': comparison[comparison.task == 'score'].sort_values('rmse').iloc[0]['model'],
    'outcome': comparison[comparison.task == 'outcome'].sort_values('f1_enroll', ascending=False).iloc[0]['model'],
    'pace': comparison[comparison.task == 'pace'].sort_values('macro_f1', ascending=False).iloc[0]['model'],
}
winners
```
