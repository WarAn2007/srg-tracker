# 04 — Final held-out evaluation

This notebook displays the one-time held-out test results for the models frozen by notebook 02. Evaluation functions live in `src/final_evaluation.py`. The notebook intentionally does not rerun test evaluation, preventing accidental repeated tuning on test results.

```python
from pathlib import Path
import sys

COLAB_ROOT = Path('/content/SRG-Tracker')
PROJECT_ROOT = COLAB_ROOT if (COLAB_ROOT / 'src').is_dir() else (Path.cwd() if (Path.cwd() / 'src').is_dir() else Path.cwd().parent)
SRC_DIR = PROJECT_ROOT / 'src'
METRICS_DIR = PROJECT_ROOT / 'reports' / 'metrics'
FIGURES_DIR = PROJECT_ROOT / 'reports' / 'figures'
PREDICTIONS_DIR = PROJECT_ROOT / 'reports' / 'predictions'
if not SRC_DIR.is_dir():
    raise FileNotFoundError(f'Cannot locate src directory from {Path.cwd()}')
sys.path.insert(0, str(PROJECT_ROOT)) if str(PROJECT_ROOT) not in sys.path else None

try:
    import pandas as pd
except ImportError as error:
    raise ImportError('Install project dependencies with: pip install -r requirements.txt') from error
```

```python
state_path = METRICS_DIR / 'final_evaluation_state.json'
metrics_path = METRICS_DIR / 'final_test_metrics.csv'
if not state_path.exists() or not metrics_path.exists():
    raise FileNotFoundError('Final evaluation is not available. Run python -m src.final_evaluation exactly once after selection is frozen.')
final_metrics = pd.read_csv(metrics_path)
final_metrics
```

```python
weekly_metrics = final_metrics[final_metrics['cutoff'].astype(str) != 'all'].copy()
weekly_metrics
```

```python
breakdown_path = METRICS_DIR / 'test_error_by_course_and_cutoff.csv'
if not breakdown_path.exists():
    raise FileNotFoundError(f'Missing detailed result table: {breakdown_path}')
pd.read_csv(breakdown_path).head(15)
```
